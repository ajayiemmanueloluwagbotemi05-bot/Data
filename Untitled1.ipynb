```python
import pandas as pd

# Load the cleaned data CSV file
df_cleaned = pd.read_csv('/content/cleaned data.csv')

# Display the first few rows to confirm successful import
display(df_cleaned.head())
```

### Chart 1: Total Revenue by Region

```python
import matplotlib.pyplot as plt

region_revenue = df_cleaned.groupby('Region')['Total_Revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(region_revenue.index, region_revenue.values, color='skyblue', edgecolor='black')
plt.title('Total Revenue by Region', fontsize=14, fontweight='bold')
plt.xlabel('Region', fontsize=12)
plt.ylabel('Total Revenue ($)', fontsize=12)
plt.tight_layout()
plt.show()
```

### Chart 2: Revenue Distribution by Product Category

```python
import matplotlib.pyplot as plt


category_revenue = df_cleaned.groupby('Product_Category')['Total_Revenue'].sum()


plt.figure(figsize=(7, 7))
plt.pie(category_revenue.values, labels=category_revenue.index, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff','#99ff99','#ffcc99'])
plt.title('Revenue Distribution by Product Category', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Chart 3: Histogram of Customer Satisfaction

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(8, 5))
plt.hist(df_cleaned['Customer_Satisfaction'].dropna(), bins=[1, 2, 3, 4, 5, 6], align='left', rwidth=0.8, color='lightgreen', edgecolor='black')
plt.title('Distribution of Customer Satisfaction', fontsize=14, fontweight='bold')
plt.xlabel('Satisfaction Score (1-5)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xticks(range(1, 6))
plt.tight_layout()
plt.show()
```

### Chart 4: Line Chart of Total Revenue Over Time

```python
import matplotlib.pyplot as plt
import pandas as pd


df_temp = df_cleaned.copy()
df_temp['Date'] = pd.to_datetime(df_temp['Date'])
revenue_by_date = df_temp.groupby('Date')['Total_Revenue'].sum().sort_index()


plt.figure(figsize=(10, 5))
plt.plot(revenue_by_date.index, revenue_by_date.values, marker='o', color='purple', linewidth=2, label='Daily Revenue')
plt.title('Total Revenue Trend Over Time', fontsize=14, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Total Revenue ($)', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
